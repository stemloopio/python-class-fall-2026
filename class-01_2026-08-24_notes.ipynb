## Why Python?
There are many different programming languages out there, but for this course, we're going with Python. But why? Before answering that question, we took a look at two sample programs written in Python.

### Guessing Game

(If you weren't present for the class, click on the box of code below and then click the Play button on the toolbar above to play the guessing game).

```python
import random

secret = random.randint(1, 100)
guesses = 0

print("I'm thinking of a number between 1 and 100.")

while True:
    guess = int(input("Your guess: "))
    guesses = guesses + 1

    if guess < secret:
        print("Too low!")
    elif guess > secret:
        print("Too high!")
    else:
        print(f"You got it in {guesses} guesses!")
        break
```

After playing through this game once, we briefly made some guesses on what each line did, but not with much detail, as we'll learn about it later.

A couple things pointed out in the code above:
- `import random` - Brings in a "library" of code that adds features that we can use
- `randint(1, 100)` - Random integer between 1 and 100
- `print()` - How to display text output to the user
- `input()` - Lets user input text
- `elif` - Means "else if"

## Typing Tests
Next we moved on to doing some typing tests by going to [Monkeytype](https://monkeytype.com) and doing 30-second tests. After each test, we wrote down our WPM (words per minute) on the smartboard. While everyone was marking there results on the board, I typed in the numbers for each student into code below.

```python
# WPM results
vidya = [19, 22, 25, 14, 23, 26]
lilikette = [26, 40, 34, 34, 34, 44]
vivek = [19, 22, 20, 25, 24, 24]
jacob = [12, 20, 22, 21, 20, 20, 22]
may = [12, 14, 18, 15, 14, 17]
danny = [77, 80, 75]
```

In the code above, I stored the data for each student into a variable named after themselves into a "list". A list is created by enclosing data between square brackets (`[` and `]`) and separating elements using a comma (`,`).

Next, we talked about what we should do with the data. Someone suggested that we calculate the average/mean of their results. To do that, we need two things: 1. how to count the number of elements in the list, and 2. add up all the elements in the list.

In Python, we can use the `len()` function to get the number of items in the list. `len` is short for "length". Python has a `sum()` function to sum up all the items in the list.

```python
print(len(vidya))
print(sum(vidya))
```

Now that we have the length and sum of the list, to computer the average, we just need to divide to two numbers:

```python
print(sum(vidya)/len(vidya))
```

One student didn't like seeing the decimals and asked if we could round the value. We most certainly can with the `round()` function that's provided by Python:

```python
print(round(sum(vidya)/len(vidya)))
```

If we wanted to calculate this for every student, it'd be tedious to type all this in. So what we can do is create a function to calculate the mean without having rewrite the same code over and over again.

```python
def mean(name):
    round(sum(name)/len(name))
    
print(mean(vidya))
print(mean(lilikette))
print(mean(vivek))
print(mean(jacob))
print(mean(may))
print(mean(danny))
```

Hmm, we ran the code, and a bunch of `None` lines were printed out. I don't program in Python frequently anymore, so I forgot that you have to explictly return a value from a function. We do this by writing `return` before the value we want to output from our function.

```python
def mean(name):
    return round(sum(name)/len(name))
    
print(mean(vidya))
print(mean(lilikette))
print(mean(vivek))
print(mean(jacob))
print(mean(may))
print(mean(danny))
```
