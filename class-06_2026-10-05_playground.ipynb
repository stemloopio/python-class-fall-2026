```python
from ipycanvas import Canvas
from ipywidgets import Image

canvas = Canvas(width=600, height=300)
canvas.line_width = 1
canvas.stroke_rect(0, 0, 600, 300)

def draw_dino(x, y):
    sprite = Image.from_file("images/larry-the-dino.png")
    canvas.draw_image(sprite, x, y, width=50)
    
def draw_flipped_dino(x, y):
    canvas.save()
    sprite = Image.from_file("images/larry-the-dino.png")
    canvas.scale(-1, 1)
    canvas.draw_image(sprite, -x - 50, y, width=50)
    canvas.restore()

for i in range(3):
    for j in range(2):
        draw_flipped_dino(50 + i*60, 50 + j*60)
canvas
```
