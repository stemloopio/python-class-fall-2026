### The volume dial

🔮 **Predict first:** what will `wave * 0.1` sound like? Run it.

```python
t = np.linspace(0, 1, RATE, endpoint=False)
wave = np.sin(2 * np.pi * 440 * t)

Audio(wave * 0.1, rate=RATE, autoplay=True)
```

😮 **It is not quieter!** Python's audio player secretly turns every sound up to the maximum. That is called *normalizing*.

Tell it to stop with `normalize=False`:

```python
Audio(wave * 0.1, rate=RATE, autoplay=True, normalize=False)
```

Now both dials together. **How fast it wiggles is pitch. How big it wiggles is volume.**

```python
freq = 440        # <-- pitch
volume = 0.5      # <-- 0 is silent, 1 is full. Try 0.2 and 0.05
seconds = 1.0

t = np.linspace(0, seconds, int(RATE * seconds), endpoint=False)
wave = volume * np.sin(2 * np.pi * freq * t)

Audio(wave, rate=RATE, autoplay=True, normalize=False)
```

## 4. Seeing the wave

This draws the first **10 milliseconds** (that is 0.01 seconds) of your sound.

One wiggle is one hill and one valley. **Count the hills.**

```python
import matplotlib.pyplot as plt

freq = 200        # <-- change me
volume = 1.0      # <-- change me

t_short = np.linspace(0, 0.01, int(RATE * 0.01), endpoint=False)

plt.figure(figsize=(9, 3))
plt.plot(t_short * 1000, volume * np.sin(2 * np.pi * freq * t_short))
plt.ylim(-1, 1)             # keeps the picture the same size so you can SEE volume
plt.xticks(range(11))
plt.grid(True)
plt.xlabel("milliseconds")
plt.show()
```

## 5. The hearing test 👂

**Set your computer volume LOW.** High pitches can hurt.

Run each cell. **Hands up if you can hear it.** Keep your hand down when you can't.

(Sounds above 11,000 wiggles per second need a faster `RATE`, so this test uses its own. Ask me why!)

```python
HI_RATE = 44100

def beep(freq, seconds=1.5):
    n = int(HI_RATE * seconds)
    t = np.linspace(0, seconds, n, endpoint=False)
    sound = 0.3 * np.sin(2 * np.pi * freq * t)
    edge = int(HI_RATE * 0.01)
    sound[:edge] *= np.linspace(0, 1, edge)
    sound[-edge:] *= np.linspace(1, 0, edge)
    return Audio(sound, rate=HI_RATE, autoplay=True, normalize=False)
```

```python
beep(8000)
```

```python
beep(12000)
```

```python
beep(15000)
```

```python
beep(17000)
```

```python
beep(19000)
```

## 6. Noise versus pattern

🔮 **Predict:** what note will this play?

```python
Audio(np.random.uniform(-1, 1, RATE), rate=RATE, autoplay=True)
```

Static. There is no note in it. Look at why:

```python
t_short = np.linspace(0, 0.01, int(RATE * 0.01), endpoint=False)
noise = np.random.uniform(-1, 1, len(t_short))
tone_440 = np.sin(2 * np.pi * 440 * t_short)

fig, (top, bottom) = plt.subplots(2, 1, figsize=(9, 5), sharex=True)
top.plot(t_short * 1000, noise)
top.set_title("Random: no pattern")
bottom.plot(t_short * 1000, tone_440)
bottom.set_title("440: the same shape over and over")
bottom.set_xlabel("milliseconds")
plt.show()
```

**Random means no pattern. Pitch *is* pattern.**

## 7. 🚨 Build your siren

A siren is a sound whose pitch **slides**. Change the numbers and press Shift+Enter.

```python
start = 200        # <-- pitch at the beginning
end = 1200         # <-- pitch at the end
seconds = 2.0      # <-- how long
volume = 0.5

n = int(RATE * seconds)
pitch = np.linspace(start, end, n)                    # slides from start to end
wave = np.sin(2 * np.pi * np.cumsum(pitch) / RATE)    # leave this line alone

Audio(wave * volume, rate=RATE, autoplay=True, normalize=False)
```

### Bonus: up and down like a real police siren

```python
seconds = 3.0
n = int(RATE * seconds)
low, high = 500, 1300

up = np.linspace(low, high, n // 2)
down = np.linspace(high, low, n // 2)
pitch = np.concatenate([up, down])      # glue the two slides together

wave = np.sin(2 * np.pi * np.cumsum(pitch) / RATE)

Audio(wave * 0.5, rate=RATE, autoplay=True, normalize=False)
```

### Bonus: UFO 🛸 (a sound made out of a sound)

```python
seconds = 3.0
n = int(RATE * seconds)
t = np.linspace(0, seconds, n, endpoint=False)

center = 800     # <-- the middle pitch
depth = 200      # <-- how far the pitch wobbles up and down
speed = 6        # <-- wobbles per second

pitch = center + depth * np.sin(2 * np.pi * speed * t)
wave = np.sin(2 * np.pi * np.cumsum(pitch) / RATE)

Audio(wave * 0.5, rate=RATE, autoplay=True, normalize=False)
```
