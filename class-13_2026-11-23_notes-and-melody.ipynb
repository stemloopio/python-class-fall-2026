```python
import numpy as np
import wave
from IPython.display import Audio

RATE = 22050

# Note name -> how many wiggles per second (Hz). We will explore this in a few minutes.
NOTES = {
    "C4": 261.63, "D4": 293.66, "E4": 329.63, "F4": 349.23,
    "G4": 392.00, "A4": 440.00, "B4": 493.88, "C5": 523.25,
    "rest": 0,
}


def tone(freq, seconds, fade=True):
    """One note as a list of numbers."""
    n = int(RATE * seconds)
    t = np.linspace(0, seconds, n, endpoint=False)
    sound = np.sin(2 * np.pi * freq * t)
    edge = int(RATE * 0.01)                    # 10 milliseconds
    if fade and n > 2 * edge:
        sound[:edge] *= np.linspace(0, 1, edge)     # fade in
        sound[-edge:] *= np.linspace(1, 0, edge)    # fade out
    return sound


def render(notes, seconds=0.4, fade=True):
    """A list of note names -> one long wave. (Look: it builds a list in a loop!)"""
    pieces = []
    for name in notes:
        pieces.append(tone(NOTES[name], seconds, fade))
    if len(pieces) == 0:
        return np.zeros(1)
    return np.concatenate(pieces) * 0.5


def play(notes, seconds=0.4, fade=True):
    return Audio(render(notes, seconds, fade), rate=RATE, normalize=False)


def save(notes, filename, seconds=0.4):
    data = (render(notes, seconds) * 32767).astype(np.int16)
    with wave.open(filename, "wb") as f:
        f.setnchannels(1)
        f.setsampwidth(2)
        f.setframerate(RATE)
        f.writeframes(data.tobytes())
    print("Saved", filename, "-", round(len(data) / RATE, 1), "seconds")
```

## 5. Building a list in a loop

```python
song = []
for note in ["C4", "E4", "G4"]:
    song.append(note)
    song.append("rest")
    print(song)
```

## 6. 🎼 Build your song

A song has parts that repeat. Change the notes in `verse` and `chorus`, change the order of the parts, change the speed.

Use **only** names from `NOTES`: `C4 D4 E4 F4 G4 A4 B4 C5 rest`

```python
verse = ["C4", "E4", "G4", "E4"]       # <-- change these notes
chorus = ["C5", "G4", "E4", "C4"]      # <-- and these
speed = 0.3                            # <-- and this

song = []
for part in [verse, chorus, verse, chorus]:    # <-- the order of your song
    for note in part:
        song.append(note)

play(song, speed)
```

**Ideas:**
- Make the chorus much longer than the verse.
- Put `"rest"` between phrases so it can breathe.
- Add a third part and add it to the order list.
