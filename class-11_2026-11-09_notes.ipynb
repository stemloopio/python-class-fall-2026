## 24 Game

In this game, you are given 24 numbers, and you may use basic math operations (`+`, `-`, `×`, `÷`) to get the result of 24. All four numbers must be used once.

For example, given the numbers of 6, 3, 5, 4, one solution is $(5-3+4)\times6 = 24$.

Another example: given 2, 2, 1, 6, one solution is $1\times(2+2)\times6 = 24$. Or $(2+1)\times(2+6) = 24$.

```python
numbers1 = [1,4,6,3]
numbers2 = [5, 9, 67]
numbers3 = [2, 7]

# First, let's remind ourselves how to do a regular for loop:
print("numbers1:")
for i in numbers1:
    print(i)

# Next, let's do a nested loop, with one for loop inside of the other:
print("numbers1 * numbers2:")
for i in numbers1:
    for j in numbers2:
        print(str(i) + '*' + str(j) + '=' + str(i*j))

# Now, let's nest one more for loop inside and multiply all combinations of the 3 lists:
print("numbers1 * numbers2 * numbers3:")
for i in numbers1:
    for j in numbers2:
        for k in numbers3:
            print(str(i) + '*' + str(j) + '*' + str(k) + '=' + str(i*j*k))

# Better way of doing the printing with format strings:
print("numbers1 * numbers2 * numbers3:")
for i in numbers1:
    for j in numbers2:
        for k in numbers3:
            print(f'{i}*{j}*{k}={i*j*k}')
```

So let's say we're given the numbers 2, 3, 4, 5 and operators +, *, -.

For the 1st tree, we'd use the operators like this: ((2 + 3) * 4) - 5

For the 2nd tree, we'd use the operators like this: 2 + (3 * (4 - 5))

For the 3rd tree, we'd use the operators like this: (2 + 3) * (4 - 5)

For the 4th tree, we'd use the operators like this: (2 + (3 * 4)) - 5

For the 5th tree, we'd use the operators like this: 2 + ((3 * 4) - 5)

Steps:

- Generate permutations for all for numbers `itertools.permutations(numbers)`
- Generate all the possible number/operator expressions
- Test each one out

```python
from itertools import permutations
OPS = ['+', '-', '*', '/']
numbers = [int(i) for i in input().split(',')]
#print(list(permutations(numbers)))

def all_tests(a, b, c, d):
    tests = []
    for op1 in OPS:
        for op2 in OPS:
            for op3 in OPS:
                # Template: f'{a} {op1} {b} {op2} {c} {op3} {d}'
                tests.append(f'(({a} {op1} {b}) {op2} {c}) {op3} {d}') # ((A.B).C).D
                tests.append(f'{a} {op1} ({b} {op2} ({c} {op3} {d}))') # A.(B.(C.D))
                tests.append(f'({a} {op1} {b}) {op2} ({c} {op3} {d})') # (A.B).(C.D)
                tests.append(f'{a} {op1} (({b} {op2} {c}) {op3} {d})') # A.((B.C).D)
                tests.append(f'({a} {op1} ({b} {op2} {c})) {op3} {d}') # (A.(B.C)).D
    return tests

for nums in permutations(numbers):
    [a, b, c, d] = nums
    for test in all_tests(a, b, c, d):
        try:
            result = eval(test)
            if abs(result - 24) < 0.001:
                print(test)
        except:
            pass
```

### Exceptions
In the code above, there's `try` and `except`. What's that? Run the line below, and you'll see a `ZeroDivisionError` error.

```python
5/0
```

For our 24 tester, we might run into a situation where we're dividing by 0, so what we can do is "catch" that error and handle it in some way. In our case, we just want to skip that permutation of numbers/ops and continue testing.

So within the `try` block, we put our code that we think might cause an error in some situations.

Then after that, we have an `except` block, which will only execute if there's an error that's caught within the `try` block. Here, we write `pass` to just let things go and continue along our way.

### Precision
Above, we also needed to add `if abs(result - 24) < 0.001`.

Why?, well for the for the numbers [3, 3, 8, 8], we end up getting 23.99999999999999 when we run the cell below. This is because when we do division sometimes, the way the resulting floating point number gets stored is not perfectly precise.

So to handle this, we write in our code that if we're really close to 24, then consider the result to be 24 and print out the test as successful.
