## Naming Conventions

Next we went over naming of variables. We brainstormed what kind of variables we need in order to describe a stuffed animal:

- Color
- Name
- Type of Animal
- Stuffed or not
- Age
- Shape
- Material

Then we talked about how the variables should be named and why. We need to have a naming convention since it makes it easier for other people to read and understand your code. It also helps yourself when you look at your own code in the future, as you may forget things.

The general rules are:
- Use all lowercase letters, with words joined by underscores (`_`)
- Must start with a letter
- Keep thing short and simple, whiile describing what the variable is used for
- Don't make it one of the Python keywords (`and, as, assert, async, await, break, class, continue, def, del, elif, else, except, finally, for, from, global, if, import, in, is, lambda, nonlocal, not, or, pass, raise, return, try, while, with, yield`)

Based on these rules, we named the brainstormed variables as follows:
- Color -> `color`
- Name -> `name`
- Type of Animal -> `animal_type` (`type_of_animal` would be a bit wordy)
- Stuffed or not -> `is_stuffed`
- Age -> `age`
- Shape -> `shape`
- Material -> `material`

## Numbers

Next, we reviewed numbers a bit and went over integers and floats.

Integers (or int for short) are.... integers, so whole numbers and also negative numbers too.

Examples: 1, 20, 0, -5

```python
age = 52
print(age)
```

We can use the `type()` function to see what type a variable is:

```python
type(age)
```

A float is short for floating point number, which is a type to represent real numbers with decimal and fractional parts.

Examples: 5.0 6.7, -3.14159

```python
puppy_age = 52.5
dog_age = 49000    # You can add an int to a float
total_ages = puppy_age + dog_age
print(total_ages)
type(total_ages)   # Adding the two comes out to be a float
```

Using the `int()` function caan be used to convert a float to an int:

```python
print(int(6))
print(int(6.2))
print(int(6.7))
type(int(6.7))
```

As you can see above, using `int()` truncates the decimal portion of the number. What about negative numbers?

```python
int(-66.7)
```

## Booleans

Booleans represent truth values and can only be `True` or `False`.

Here are some of the operators you can use for comparison:
- `<` Less than
- `>` Greater than
- `==` Equals (note the double equal signs, since single equal sign is variable assignment)
- `!=` Not equal
- `<=` Less than or equal to
- `>=` Greater than or equal to

```python
5 < 6
```

```python
5 > 6
```

```python
5 == 5
```

```python
5 == 6
```

```python
5 != 6
```

```python
5 <= 5
```

```python
5 >= 6
```

You can also compare strings using these same operators:

```python
'car' == 'cat'
```

```python
# This fails since you can't assign to a string
'catcat' = 5
```

```python
'c'*3 == 'ccc'
```

You can also cast a string to an int using the `int()` function. An example of its use is when reading in a numerical input:

```python
'5' == 5
```

```python
int('5')
```

```python
int('5') == 5
```

## Arrays

Last time, we stored some data in arrays, so we'll touch on them some more again.

```python
speeds = [3, 6, 7, 8, 11]
print(speeds)
```

Arrays have some methods we can use on them to figure things out like, length, sum, max value, etc.

```python
print(len(speeds))
print(max(speeds))
print(min(speeds))
print(sum(speeds))
```

We can also access elements of the array by using brackets `[ ]`.

The first element is accessed by using `[0]`, the next with `[1]`, and so forth.
This is known as 0-indexing, so for getting the elements in `speeds`, we have to use the following indicies:

| index  | 0 | 1 | 2 | 3 | 4 |
| --- | --- | --- | --- | --- | --- |
| speeds | 3 | 6 | 7 | 8 | 9 |

We can also get the last element of the array by using the index of -1 as follows:

```python
speeds[-1]
```

The 2nd to last element can be grabbed with index of -2:

```python
speeds[-2]
```

Python checks if you use an index that is largers than the size of the array:

```python
speeds[11]
```

Lastly, you can add elements to the  end of the array by using `.append()`:

```python
speeds.append(2)
print(speeds)
```

```python
speeds.append(200)
speeds.append(6)
print(speeds)
```
