### Recreating the Game

Now that we've played through the game a couple times, let's create our own version of the game. To keep things simple, we'll change out clicking of items with typing in of numbers. So for example, if we're on Novice Level, there were 6 buttons, so we'll just type in a number in between 1 and 6 instead. Likewise, numbers will be flashed on the screen for us to memorize and input.

So what do we need to do to make this game?

- Need to generate the pattern to follow (random number generation)
- Need to progress from displaying/inputting 1 item in the pattern to however long the pattern is
- Not in the original game, but tracking number of errors player made
- Later on, replace numbers with colorful buttons/items

## Breakdown of Program

- Configure game parameters
  - Number of buttons
  - Length of pattern
- Initialize state (State is keeping track of where we are in playing puzzle)
  - Current pattern progress
- Play game
  - For each item in pattern, play a round until successful

First, let's set up our game paramester and initialize state:

```python
# Set up game parameters
num_buttons = 6
pattern_length = 4

# Initialize state
current_round = 1
```

Next, we need to generate the pattern. We know how to generate a single random number in Python using `random.randint(1, num_buttons)`, but how do we generate more than 1? To do this, we need to create an list of numbers. We'll start off by creating an empty list and then adding to it:

```python
import random
pattern = []   # Create an empty list
for i in range(pattern_length):
    pattern.append(random.randint(1, num_buttons))
    print(pattern)   # Print out current list
```

What's this `for` thing that is in line 3 above? It's what's called a for loop. Let's break it down. First, `range(pattern_length)` generates the following:

```python
range(pattern_length)
```

Basically, this means `(0, 1, 2, 3)` (a series of 4 numbers starting from 0). The actual numbers in this sequence aren't important to us in this case, just the amount of numbers.

Next, the `i` in the expression is used to grab one element at a time from the range item.

```python
for i in range(pattern_length):
    print(i)
```

```python
for i in [5, 6, 1, 3, 0]:
    print(i)
```

Now that we've generated the pattern, let's setup a function to play a round of the game:

```python
from IPython.display import clear_output
import time

def show_pattern(round_number):
    print('Here is the pattern so far:')
    time.sleep(1)
    clear_output(wait=True)
    for n in pattern[:round_number]:
        print(n)
        time.sleep(1)
        clear_output(wait=True)
        print('')
        time.sleep(0.5)
        clear_output(wait=True)

def guess_pattern(round_number):
    for i in range(round_number):
        clear_output(wait=True)
        print('Type next number in pattern:')
        guess = int(input())
        if guess != pattern[i]:
            return False
    return True

def play_round(round_number):
    while True:
        clear_output(wait=True)
        show_pattern(round_number)

        if guess_pattern(round_number):
            print(f'Round {round_number} completed')
            return
            
        print('Wrong guess, try pattern again')
        time.sleep(2)

    # We guessed all the numbers in the current round of the pattern correctly
    print('Round completed')
    return

# Test things out
#play_round(2)
```

```python
def play_game():
    for i in range(pattern_length):
        play_round(i + 1)
    print('Congratulations, you beat the puzzle!')

play_game()
```
